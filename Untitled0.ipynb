```python
import numpy as np

# Input point
x, y = 100, 50

# Rotation settings
angle = 90
direction = 1   # 1 = CCW, -1 = CW

P = np.array([x, y], dtype=float)

print(f"Original Point: {P}")
print(f"Rotation Angle: {angle}°")
print(f"Direction: {'Counterclockwise' if direction == 1 else 'Clockwise'}")
```

```python
def rotation_matrix(angle, direction=1):
    theta = np.radians(direction * angle)

    return np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])

Q = rotation_matrix(angle, direction)

# Transform point
P_rotated = Q @ P

print("Rotation Matrix Q:")
print(np.round(Q, 4))

print("\nRotated Point P' = QP:")
print(np.round(P_rotated, 4))
```

```python
QTQ = Q.T @ Q
det_Q = np.linalg.det(Q)

print("QᵀQ:")
print(np.round(QTQ, 4))

print(f"\ndet(Q) = {det_Q:.4f}")

print("\nOrthogonal:", np.allclose(QTQ, np.eye(2)))
print("Proper Rotation:", np.isclose(det_Q, 1))
```

```python
original_distance = np.linalg.norm(P)
rotated_distance = np.linalg.norm(P_rotated)

original_quadratic = P.T @ P
rotated_quadratic = P_rotated.T @ P_rotated

print("Distance:")
print(f"Original : {original_distance:.4f}")
print(f"Rotated  : {rotated_distance:.4f}")

print("\nQuadratic Form:")
print(f"PᵀP   = {original_quadratic:.4f}")
print(f"P'ᵀP' = {rotated_quadratic:.4f}")

print("\nDistance Preserved:",
      np.isclose(original_distance, rotated_distance))

print("Quadratic Form Invariant:",
      np.isclose(original_quadratic, rotated_quadratic))
```

```python
# Rotate twice
Q2 = Q @ Q
P2 = Q @ P_rotated

print("DOUBLE ROTATION")
print("=" * 35)

print("Q²:")
print(np.round(Q2, 4))

print("\nAfter 1st rotation:")
print(np.round(P_rotated, 4))

print("\nAfter 2nd rotation:")
print(np.round(P2, 4))

if np.isclose(abs(angle), 90):
    print("\n✓ Two 90° rotations = 180° rotation")
    print("✓ Q² = -I:", np.allclose(Q2, -np.eye(2)))

print("\nFINAL BACKEND STATUS")
print("=" * 35)
print("Orthogonal           :", np.allclose(QTQ, np.eye(2)))
print("det(Q) = +1          :", np.isclose(det_Q, 1))
print("Distance preserved   :", np.isclose(original_distance, rotated_distance))
print("Quadratic invariant   :", np.isclose(original_quadratic, rotated_quadratic))
```
